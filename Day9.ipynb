```python
import numpy as np  
x = np.arange(4)  
y1 = [10, 15, 7, 12]  
y2 = [8, 14, 9, 10] 

width=0.4

plt.bar(x-(width/2),y1,width,edgecolor='black',linewidth=2,label="Group1")
plt.bar(x+(width/2),y2,width,edgecolor='black',linewidth=2,label="Group2")
plt.xticks(x,['A','B','C','D'])
plt.xlabel("categories")
plt.ylabel("values")
plt.legend()
plt.show()
```

```python
#c specifies the color or color values of the data points, while cmap specifies the color mapping scheme used to convert numerical values into colors.
import numpy as np
value=np.array([1,2,3,4,5])
x = [1, 2, 3, 4, 5]  
y = [2, 4, 1, 3, 5] 
plt.scatter(x,y,c=value,cmap="viridis")
plt.colorbar(label='Color Scale') 
plt.title("color map plot")
plt.show()
```

```python
x = [1, 2, 3, 4, 5]  
y = [2, 4, 1, 3, 5] 
y2 = [5, 3, 4, 2, 1]  
plt.scatter(x,y,label="Group1",c="blue")
plt.scatter(x,y2,label="Group2",c="green")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
```

```python
x = [1, 2, 3, 4, 5]  
y = [5, 7, 6, 8, 7]  
plt.scatter(x, y)  
# Annotate each point  
for i in range(len(x)):  
    plt.text(x[i]+0.1, y[i]+0.1, f"({x[i]}, {y[i]})")
```

```python
sizes = [30, 20, 25, 25]  
labels = ['A', 'B', 'C', 'D']  

plt.pie(sizes,labels=labels,autopct="%1.1f%%")
plt.show()
```

```python
color=["green","yellow","purple","blue"]
plt.pie(sizes,labels=labels,autopct="%1.1f%%",colors=color)
plt.show()
```

```python
import matplotlib.pyplot as plt
data = [1, 2, 2, 3, 3, 3, 4, 4, 5, 5, 5, 5]  
plt.hist(data,bins=5,color="skyblue",edgecolor="black")
plt.xlabel("x_axis")
plt.ylabel("y_axis")
plt.title("Histogram plot")
plt.grid(axis="y",linestyle="--",alpha=0.7)
plt.show()
```

```python
data = [1, 2, 2, 3, 3, 3, 4, 4, 5, 5, 5, 5]  
data2 = [2, 3, 3, 4, 4, 5, 5, 5, 6]  
plt.hist(data,bins=5,alpha=0.5,label='Group1',color='blue',edgecolor='black')
plt.hist(data2,bins=5,alpha=0.5,label='Group2',color='red')
plt.axvline(x=4,color='black',linestyle='--',linewidth=2)
plt.show()
```

```python
data2 = [2, 3, 3, 4, 4, 5, 5, 5, 6]  
plt.hist(data2,bins=5,alpha=0.7,orientation="horizontal",color='red',edgecolor='black',linewidth=2)
plt.xlabel("Freq")
plt.ylabel("Bins")
plt.title("Horizontal Histogram")
plt.show()
```

```python
x = [1, 2, 3, 4, 5]  
y1 = [1, 2, 3, 4, 5]  
y2 = [2, 1, 2, 1, 2]  
plt.stackplot(x,y1,y2,labels=['Group1','Group2'],colors=['blue','red'])
plt.xlabel("x-axis")
plt.ylabel('y-axis')
plt.legend(loc="upper left")
plt.show()
```

```python

x = [1, 2, 3, 4, 5]  
y1 = [2, 3, 5, 7, 11]  
y2 = [1, 4, 2, 5, 3]  
plt.subplot(1,2,1)
plt.plot(x,y1,label="Graph1",color='blue')
plt.xlabel("x_label")
plt.ylabel("y1_label")

plt.subplot(1,2,2)
plt.plot(x,y2,label="Graph2",color='red')
plt.xlabel("x_label")
plt.ylabel("y2_label")

plt.tight_layout()
plt.show()
```

```python
x = [1, 2, 3, 4, 5]  
y = [2, 3, 5, 7, 11]  
fig,ax=plt.subplots()

ax.plot(x,y,label='prime number',color='red',linestyle='--',marker='o')
ax.set_xlabel("x-axis")
ax.set_ylabel("y-axis")
ax.set_title("subplots")
ax.legend()
plt.show()
```

```python
x = [1, 2, 3, 4, 5]  
y1 = [2, 3, 5, 7, 11]  
y2 = [1, 4, 2, 5, 3] 
fig,axs=plt.subplots(2,2,figsize=(8,6))

axs[0,0].plot(x,y1)
axs[0,0].set_title("top_left")

axs[0,1].bar(['A','B','C'], [3,5,2])  
axs[0,1].set_title('top_right')

axs[1,0].scatter(x,y2)
axs[1,0].set_title('bottom_left')

axs[1,1].hist([1,2,2,3,3,3,4])  
axs[1,1].set_title("bottom_right")

plt.tight_layout()
plt.show()
```
